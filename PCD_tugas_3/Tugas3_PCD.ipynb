NO.1 rgb ke grayscale

```python
import cv2
import matplotlib.pyplot as plt

# 1. Baca gambar (format BGR dari OpenCV)
img = cv2.imread('/content/images.jpg')

# 2. Ubah ke RGB (agar warna benar saat ditampilkan)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 3. Konversi ke Grayscale
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 4. Tampilkan hasil
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("RGB Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

2. Graycale ke biner

```python
import cv2
import matplotlib.pyplot as plt

# Baca gambar grayscale
img_gray = cv2.imread('/content/images.jpg', 0)

# Threshold (misal 128)
_, img_binary = cv2.threshold(img_gray, 128, 255, cv2.THRESH_BINARY)

# Tampilkan
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img_binary, cmap='gray')
plt.title("Binary")
plt.axis("off")

plt.show()
```

3. Grayscale ke mbit

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Baca gambar grayscale
img_gray = cv2.imread('/content/images.jpg', 0)

def quantize(image, m_bit):
    levels = 2 ** m_bit
    step = 256 // levels
    return (image // step) * step

# Contoh m-bit
img_2bit = quantize(img_gray, 2)
img_3bit = quantize(img_gray, 3)
img_4bit = quantize(img_gray, 4)

# Tampilkan
plt.figure(figsize=(8,6))

plt.subplot(2,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original (8-bit)")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(img_2bit, cmap='gray')
plt.title("2-bit (4 level)")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(img_3bit, cmap='gray')
plt.title("3-bit (8 level)")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(img_4bit, cmap='gray')
plt.title("4-bit (16 level)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

4. grightness

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/images.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

brightness = cv2.convertScaleAbs(img, alpha=1, beta=50)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Before")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(brightness)
plt.title("After (Brightness)")
plt.axis("off")

plt.show()
```

5. contrast

```python
import cv2
import matplotlib.pyplot as plt

# Baca gambar
img = cv2.imread('/content/images.jpg')

# Cek apakah gambar terbaca
if img is None:
    print("Gambar tidak ditemukan!")
else:
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Tambah contrast (alpha)
    contrast = cv2.convertScaleAbs(img, alpha=1.5, beta=0)

    # Tampilkan BEFORE & AFTER
    plt.figure(figsize=(10,5))

    # Before
    plt.subplot(1,2,1)
    plt.imshow(img)
    plt.title("Before (Original)")
    plt.axis("off")

    # After
    plt.subplot(1,2,2)
    plt.imshow(contrast)
    plt.title("After (Contrast +)")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

6. graycale to histogram

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/images.jpg')

# Cek gambar
if img is None:
    print("Gambar tidak ditemukan!")
else:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    plt.figure(figsize=(12,4))

    # BEFORE (RGB)
    plt.subplot(1,3,1)
    plt.imshow(img_rgb)
    plt.title("Before (RGB)")
    plt.axis("off")

    # AFTER (Grayscale)
    plt.subplot(1,3,2)
    plt.imshow(gray, cmap='gray')
    plt.title("After (Grayscale)")
    plt.axis("off")

    # Histogram (After)
    plt.subplot(1,3,3)
    plt.hist(gray.ravel(), bins=256)
    plt.title("Histogram (Grayscale)")

    plt.tight_layout()
    plt.show()
```

7. operasi histogram equalization

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/images.jpg', 0)

# Cek gambar
if img is None:
    print("Gambar tidak ditemukan!")
else:
    hist_eq = cv2.equalizeHist(img)

    plt.figure(figsize=(12,4))

    # BEFORE
    plt.subplot(1,3,1)
    plt.imshow(img, cmap='gray')
    plt.title("Before (Grayscale)")
    plt.axis("off")

    # AFTER
    plt.subplot(1,3,2)
    plt.imshow(hist_eq, cmap='gray')
    plt.title("After (Equalization)")
    plt.axis("off")

    # Histogram Before vs After
    plt.subplot(1,3,3)
    plt.hist(img.ravel(), bins=256, alpha=0.5, label='Before')
    plt.hist(hist_eq.ravel(), bins=256, alpha=0.5, label='After')
    plt.title("Histogram Comparison")
    plt.legend()

    plt.tight_layout()
    plt.show()
```
